## Day 3 – Data Cleaning

### Outlier Detection


```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

print(numerical_columns)
```

```python
import matplotlib.pyplot as plt

# Create boxplots for all numerical columns
df[numerical_columns].plot(
    kind='box',
    subplots=True,
    layout=(7, 4),
    figsize=(15, 20)
)

plt.tight_layout()
plt.show()
```

## Outlier Analysis

Outliers were identified using boxplots for all numerical features.

A few columns, such as **MonthlyIncome**, **TotalWorkingYears**, **YearsAtCompany**, **YearsInCurrentRole**, **YearsSinceLastPromotion**, and **YearsWithCurrManager**, contain values outside the interquartile range.

These observations represent employees with higher salaries, longer experience, or longer tenure rather than data entry errors. Therefore, the outliers were retained to preserve the integrity and completeness of the dataset.

```python
# Detect outliers using the IQR method

Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)
IQR = Q3 - Q1

outlier_count = (
    (df[numerical_columns] < (Q1 - 1.5 * IQR)) |
    (df[numerical_columns] > (Q3 + 1.5 * IQR))
).sum()

print(outlier_count)
```

## **FEATURE** **ENGINEERING**

```python
# Create Total Satisfaction Score
df['TotalSatisfactionScore'] = (
    df['EnvironmentSatisfaction'] +
    df['JobSatisfaction'] +
    df['RelationshipSatisfaction'] +
    df['WorkLifeBalance']
)

df[['TotalSatisfactionScore']].head()
```

```python
# Experience before joining current company
df['YearsWorkedOutsideCompany'] = (
    df['TotalWorkingYears'] -
    df['YearsAtCompany']
)

df[['YearsWorkedOutsideCompany']].head()
```

## Feature Engineering

Two new features were created to enrich the dataset:

- **TotalSatisfactionScore**: Combines Environment Satisfaction, Job Satisfaction, Relationship Satisfaction, and Work-Life Balance into a single overall satisfaction score.

- **YearsWorkedOutsideCompany**: Calculates the employee's work experience before joining the current company by subtracting YearsAtCompany from TotalWorkingYears.

These engineered features provide additional information while preserving the original dataset.

# **NORMALIZATION**

```python
# Create a copy of the cleaned dataset
df_normalized = df.copy()

from sklearn.preprocessing import MinMaxScaler

numerical_columns = df_normalized.select_dtypes(include=['int64', 'float64']).columns

scaler = MinMaxScaler()

df_normalized[numerical_columns] = scaler.fit_transform(
    df_normalized[numerical_columns]
)
```
